```python
import matplotlib.pyplot as plt
import numpy as np

# Data
models = ['GPT-4', 'BERT Large', 'GPT-3', 'Our Model', 'GPT-2']
scores = [95.3, 48.0, 33.7, 33.0, 29.6]

# Sort data in descending order
sorted_indices = np.argsort(scores)[::-1]
models = [models[i] for i in sorted_indices]
scores = [scores[i] for i in sorted_indices]

# Create the plot
plt.figure(figsize=(10, 6))
bars = plt.bar(models, scores, color='#3B82F6')

# Customize the plot
plt.title('HellaSWAG Benchmark Scores', fontsize=14, pad=20)
plt.ylabel('Score (%)', fontsize=12)
plt.ylim(0, 100)  # Set y-axis range from 0 to 100

# Add value labels on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height}%',
             ha='center', va='bottom')

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Add grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Adjust layout to prevent label cutoff
plt.tight_layout()

plt.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Data
models = ['GPT-4', 'BERT Large', 'GPT-3', 'Our Model', 'GPT-2']
scores = [95.3, 48.0, 33.7, 33.0, 29.6]

# Sort data in descending order
sorted_indices = np.argsort(scores)[::-1]
models = [models[i] for i in sorted_indices]
scores = [scores[i] for i in sorted_indices]

# Create the plot
fig = go.Figure(data=[
    go.Bar(
        x=models,
        y=scores,
        text=[f'{score}%' for score in scores],
        textposition='auto',
        marker_color='#3B82F6'
    )
])

# Update layout
fig.update_layout(
    title={
        'text': 'HellaSWAG Benchmark Scores',
        'y':0.95,
        'x':0.5,
        'xanchor': 'center',
        'yanchor': 'top',
        'font': {'size': 24}
    },
    yaxis_title='Score (%)',
    yaxis_range=[0, 100],
    showlegend=False,
    plot_bgcolor='white',
    yaxis=dict(
        gridcolor='rgba(0,0,0,0.1)',
        zerolinecolor='rgba(0,0,0,0.2)'
    )
)

# Display the plot
fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Data
models = ['GPT-4', 'BERT Large', 'GPT-3', 'Our Model', 'GPT-2']
scores = [95.3, 48.0, 33.7, 33.0, 29.6]

# Sort data in descending order
sorted_indices = np.argsort(scores)[::-1]
models = [models[i] for i in sorted_indices]
scores = [scores[i] for i in sorted_indices]

# Create the plot
fig = go.Figure(data=[
    go.Bar(
        x=models,
        y=scores,
        text=[f'{score}%' for score in scores],
        textposition='auto',
        marker=dict(
            color=['#3B82F6', '#60A5FA', '#93C5FD', '#BFDBFE', '#DBEAFE'], # Gradient effect
            line=dict(color='rgba(0,0,0,0.3)', width=1),
        ),
        hoverinfo='x+y',
        textfont=dict(size=14),
        width=0.5,
    )
])

# Update layout
fig.update_layout(
    title={
        'text': 'HellaSWAG Benchmark Scores',
        'y': 0.95,
        'x': 0.5,
        'xanchor': 'center',
        'yanchor': 'top',
        'font': {'size': 26, 'family': 'Arial, sans-serif', 'color': '#333'}
    },
    xaxis=dict(
        title='Models',
        titlefont=dict(size=16),
        tickfont=dict(size=14),
        showline=True,
        linecolor='rgba(0,0,0,0.3)',
        mirror=True
    ),
    yaxis=dict(
        title='Score (%)',
        titlefont=dict(size=16),
        tickfont=dict(size=14),
        range=[0, 100],
        gridcolor='rgba(0,0,0,0.1)',
        zerolinecolor='rgba(0,0,0,0.2)'
    ),
    plot_bgcolor='rgba(250, 250, 250, 1)',
    margin=dict(l=60, r=40, t=80, b=60),
    showlegend=False
)

# Display the plot
fig.show()
```
