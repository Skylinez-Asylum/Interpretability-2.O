```python
from inference import inference
from config import GPTConfig
from model import GPT
```

```python
import torch

def remove_prefix(state_dict, prefix):
    """
    Remove a prefix from state_dict keys.
    """
    return {k[len(prefix):] if k.startswith(prefix) else k: v for k, v in state_dict.items()}

def load_model(checkpoint_path):
    # Load the checkpoint
    checkpoint = torch.load(checkpoint_path, map_location='cuda')

    # Extract the model's state_dict
    model_state_dict = checkpoint['model']

    # Remove the "_orig_mod." prefix from the keys
    model_state_dict = remove_prefix(model_state_dict, "_orig_mod.")

    # Initialize the new model with the same architecture
    new_model = GPT(GPTConfig)  # Assuming GPT and GPTConfig are defined
    
    # Load the state_dict directly into the new model
    new_model.load_state_dict(model_state_dict)
    
    return new_model

```
