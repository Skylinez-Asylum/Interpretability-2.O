```python
from model import GPT, GPTConfig  # Your local model code
import torch
from transformers import AutoTokenizer

# Load the model
config = GPTConfig()
model = GPT(config)
model.load_state_dict(torch.load("/home/arjun/Desktop/GitHub/Interpretability-2.O/customGPT2/save_states/12headNFT900k.pt", weights_only=False))
model.eval()

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# Move to device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Prepare input
prompt = "My name is"
inputs = tokenizer(prompt, return_tensors="pt").to(device)
input_ids = inputs["input_ids"]

# Manual generation loop
max_length = 300
with torch.no_grad():
    for _ in range(max_length - input_ids.shape[1]):
        logits, _ = model(input_ids, return_logits=True)  # Get logits from the last token
        next_token_logits = logits[:, -1, :]  # Take logits of the last token
        probs = torch.softmax(next_token_logits, dim=-1)  # Convert to probabilities
        next_token = torch.multinomial(probs, num_samples=1)  # Sample next token
        input_ids = torch.cat([input_ids, next_token], dim=1)  # Append to sequence

# Decode and print
generated_text = tokenizer.decode(input_ids[0], skip_special_tokens=True)
print(generated_text)
```
